### Naive

```python
arr = []

for i in range(10) :
    if(i%2 == 0) :
        arr.append(i)
print(arr)
```

### Real

```python
arr = [i + 10 for i in range(100) if i%2!=0]
print(arr)
```

### String

```python
## Naive
fruits = ['Lichi', 'mango', 'apple']
up = []

for fruit in fruits :
    up.append(fruit.upper())

print(up)
```

```python
# Main
fruits = ['Lichi', 'mango', 'apple']
up = [fruit.lower() for fruit in fruits]

print(up)
```
