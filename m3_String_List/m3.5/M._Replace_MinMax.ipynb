```python
n = int(input())
arr = input()
arr = arr.split()

mn = int(arr[0]) ; min_i = 0 ; mx = int(arr[0]) ; max_i = 0

for i in range(n) :
    arr[i] = int(arr[i])
    if(arr[i] < mn) :
        mn = arr[i]
        min_i = i
    if(arr[i] > mx) :
        mx = arr[i]
        max_i = i

tmp = arr[min_i]
arr[min_i] = arr[max_i]
arr[max_i] = tmp

for i in range(n) :
    print(arr[i], end=" ")
```
