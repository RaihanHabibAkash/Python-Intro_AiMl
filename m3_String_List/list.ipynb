### List

```python
num = [10,20,30,40,50]
string_list = ["Raihan", "Habib", "Akash"]
add_2d = [[10,20], [100,200], [1000,2000]]
mix = [19, "Raihan", 22.3]

print(num[0])
print(string_list)
print(add_2d)
print(type(mix[0]), type(mix[1]), type(mix[2]))
```

### Access & Update

```python
print(num[2])

num[2] = 100 # muteable
print(num[2])
```

### Slice

```python
new_num = num[0:3]
print(new_num)
```

### Methods

```python
print(num)
```

```python
num.append(10) # push_back
print(num)

num.pop() # pop_back
print(num)

num.insert(2,10)
print(num)

num.remove(10)
print(num)
```

### Sort & Reverse

```python
nums = [50,49,29,199,20]
nums.reverse()
print(nums)

nums.sort()
print(nums)

nums.sort(reverse=True)
print(nums)
```
