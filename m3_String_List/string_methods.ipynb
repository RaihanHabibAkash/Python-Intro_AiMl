### Find method

```python
message = "Raihan Habib Akash Habib"

print("Akash" in message)
# returns bool
print(type("Akah" in message))

#index ber kora first ar
i = message.find("Habib")
print(i)

# last r theke search shuru korbe
i = message.rfind("Habib")
print(i)
```

### Len Method

```python
print(len(message))
```

### LowerCase, UpperCase, Capital method

```python
low = message.lower()
upp = message.upper()
cap = message.capitalize()
print(low)
print(upp)
print(cap)
```

### Count Substr, replace

```python
msg = "hlw ml from ml to ml"
print(msg.count("ml"))
print(msg.replace("ml", "Ai/Ml"))
```

### Spliting, Join

```python
msg = "Raihan Habib Akash"
print(msg.split())

msg = "Raihan-Haib-Akash"
val = msg.split("-")
print(val)

print(",".join(val))
```

### Formated String

```python
name = "Akash"
roll = 45
height = 5.7892

msg = f"Hello my name is {name.upper()}. My roll is {roll-10}. My height is {height:.3}"
print(msg)

per = 4.8989823
print(f"the percentage is {per:.2%}")
```
