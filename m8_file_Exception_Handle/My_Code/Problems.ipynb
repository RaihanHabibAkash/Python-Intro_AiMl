## Practice 1 :

Get the text from sample.txt , and count the following


Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce

with open("./sample_data/sample.txt", 'r') as file :
  string_list = file.readlines()

  # File lines
  lines = len(string_list)
  print(lines)

  # Numbers of words
  words_count = list(map(lambda x: len(x.split()), string_list))
  total_words = reduce(lambda x,y: x+y, words_count)
  print(total_words)

  # chars count
  #Clean
  string_list = list(map(str.strip, string_list))
  string_list = list(map(lambda x: x.replace(" ", ""), string_list))
  # now count
  char_count = list(map(lambda x: len(x), string_list))
  total_chars = reduce(lambda x,y : x+y, char_count)
  print(total_chars)

with open("./sample_data/counter_of_string.txt", 'w') as file :
  file.write(f"Total Lines: {lines}\nTotal Words: {total_words}\nTotal Chars: {total_chars}")
```
