## Doc String

```python
def adder(a, b) :
    """
    return a+b
    
    a-> int
    b-> int
    """
    return a+b

print(help(adder))
print(help(print))
```

### Parameter

```python
def welcome_msg(user="None") :
    print(f"Hello {user}")

welcome_msg("Akash")
welcome_msg()
```

### Return

```python
def adder(a, b) :
    return a+b
print(adder(5,4) * 10)
```

### Positional Arguments

```python
# Normal function ar moto serial maintain
def adder(a, b, c) :
    print(a,b,c)
adder(5,4,8)
```

### Variable sized Arguments

```python
def hlw(*args) :
    print(args)
hlw(5,4,8,10)
```

### KeyWord Arguments

```python
def hlw(a,b,c) :
    print(a,b,c)
hlw(a = 10, c = 30, b = 20)
```

### Variable sized keywords Arguments

```python
def hlw(**kwargs) :
    print(kwargs)
hlw(a = 10, c = 30, b = 20)

def hlw(**kwargs) :
    for key,val in kwargs.items() :
        print(key, val)
hlw(a = 10, c = 30, b = 20)
```

### Unzip

```python
def vals(a,b) :
    return [a,b]

x,y = vals(10,20)
print(y,x)
```

### Iterator

```python
s = {10,20,30,40}

s_it = iter(s)

print(s_it)
print(next(s_it))
print(next(s_it))
print(next(s_it))
print(next(s_it))
```

### Generator

```python
def data_loader(chunk_size, lst) :
    for i in range(1, len(lst), chunk_size) :
        yield lst[i:i+chunk_size]

lst = [x for x in range(100)]
x = data_loader(5,lst)
print(type(x), x)

print(next(x))
print(next(x))
```
