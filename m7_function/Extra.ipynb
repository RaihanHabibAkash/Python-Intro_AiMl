### D. The Model Registry

```python
def mod(name, layers, lr=0.01) :
    return f"Model: {name}, Layers: {layers}, LR: {lr}"

n = int(input())
while n :
    inp = input().split()

    if inp[0] == '1' :
       print(mod(inp[1], int(inp[2])))
    elif inp[0] == '2' :
        print(mod(name = inp[1], layers = int(inp[2])))
    else :
        print(mod(inp[1], inp[2], float(inp[3])))
    n -= 1
```

### F. GPU Can Eat Only Three Images

```python
def data_load(lst, batch_size) :
    if(batch_size < 1) :
        return
    for i in range(0, len(lst), batch_size) :
        yield lst[i:i+batch_size]
def mean(val) :
    sum = 0
    for x in val:
        sum += int(x)
    return sum / len(val)
    
n,batch_size = list(map(int, input().split()))
lst = list(map(int, input().split()))

data = data_load(lst, batch_size)

for val in data :
    mn = mean(val)
    print(f"{val} mean={mn}")
```
