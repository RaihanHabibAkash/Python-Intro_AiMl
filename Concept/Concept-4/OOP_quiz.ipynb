```python
# how the pillars are used in this code all together?

from abc import ABC, abstractmethod


class Person(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def introduce(self):
        pass


class Student(Person):
    def __init__(self, name, cgpa):
        super().__init__(name)
        self.__cgpa = cgpa

    @property
    def cgpa(self):
        return self.__cgpa

    def introduce(self):
        print(f"I am {self.name}, a student.")

    def show_result(self):
        print(f"CGPA: {self.__cgpa}")


class Teacher(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def introduce(self):
        print(f"I am {self.name}, a teacher.")

    def teach(self):
        print(f"Teaching {self.course}")


student = Student("Nila", 3.85)
teacher = Teacher("Rahim", "Python")

people = [student, teacher]

for person in people:
    person.introduce()

student.show_result()
teacher.teach()
```

```python
class Player:
    def __init__(self, name):
        self.name = name

    def play(self):
        print(f"{self.name} is playing")


p1 = Player("A")
p2 = Player("B")

p1.name = "C"

p1.play()
p2.play()
```

```python
class Parent:
    def greet(self):
        print("Hello from parent")


class Child(Parent):
    def greet(self):
        print("Hello from child")


obj = Child()
obj.greet()
```
