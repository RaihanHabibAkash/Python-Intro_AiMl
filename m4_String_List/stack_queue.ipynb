### List as Stack

```python
stack = []

stack.append(10)
stack.append(20)
stack.append(30)
stack.append(40)
stack.append(50)


print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack.pop())
print(stack.pop())
```

### List as Queue

```python
queue = []

queue.append(10)
queue.append(20)
queue.append(30)
queue.append(40)
queue.append(50)


print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))
```
