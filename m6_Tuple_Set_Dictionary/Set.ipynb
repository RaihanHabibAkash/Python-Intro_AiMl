# Set Mutable

### Decleare

```python
st = {4,2,3,4,2,2,5,1}

print(st, type(st))
```

### Empty set

```python
st1 = {} # Can't decleare empty set like this
st2 = set()
lst = []
tup = ()

print(type(st1))
print(type(st2))
print(type(lst))
print(type(tup))
```

### Access & Iterate

```python
st = {10,"Aks", False, 2.5, 20}

if "Aksj" in st :
    print("YES")
else :
    print("NO")

for element in st :
    print(element, end=" ")
```

### Methods

```python
sub = {1,2,3}

sub.add(10)
print(sub)

sub.pop()
print(sub)

sub.remove(2)
print(sub)
```

### Math methods

```python
a = {1,2,3} ; b = {2,3,4}

print(a.union(b))
print(a.intersection(b))

a = {1,2,3} ; b = {4,5,6}
print(a.isdisjoint(b))

a = {1,2,3} ; b = {3,2,1,4,5,6}
print(a.issubset(b))
```
