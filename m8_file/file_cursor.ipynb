```python
file = open("./sample_data/sample.txt", 'r')

con = file.readlines() # pura file list a rakhbe (new line '/n' ar bhittite)
con = list(map(str.strip, filter(lambda x: len(x) > 1, con)))
print(con)

file.close()
```

Way 2 of open File (don't need to close() manually)

```python
with open("./sample_data/sample.txt", 'r') as file : # Objest create kore ja file a rakhtasi
  print(file.readlines())

print(file.closed) # auto close hoa jabe
```

```python
with open("./sample_data/sample.txt", 'r') as file :
  for line in file :
    print(line, end='')
```

Write

```python
# Creates new file if there is no one
with open("./sample_data/text.txt", 'w') as file :
  file.write("World is not good\n")
  file.write("You are good\n")

# Over write the file, if there is one exists
with open("./sample_data/text.txt", 'w') as file :
  file.write("Hello bro\n")
  file.write("I am Akash\n")
```

```python
# Append
with open("./sample_data/text.txt", 'a') as file :
  file.write("LLLLLLLLLLLLLLLLLLLLLLLLLLL\n")
```

```python
# Add List of strings
words = ["Hlw", "Bro", "NO"]
with open("./sample_data/text.txt", 'w') as file :
  file.writelines(words)
```

File pointer / Cursor

```python
with open("./sample_data/text.txt", 'r') as file :
  print(file.tell())
  print(file.read())
  print(file.tell())

  # Last index a pora jabe na
  print(file.read())
```

```python
with open("./sample_data/text.txt", 'r') as file :
  print(file.tell())
  print(file.read(5))
  print(file.tell())

  print(file.read())
  print(file.tell())
```

```python
with open("./sample_data/text.txt", 'r') as file :
  print(file.tell())
  print(file.read(5))
  print(file.tell())

  file.seek(0)
  print(file.tell())
  print(file.read())
```
