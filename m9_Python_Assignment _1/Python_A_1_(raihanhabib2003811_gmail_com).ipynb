## Ans of the Question  **1**

```python
lst = list(map(int, input().split()))
lst = list(filter(lambda x: x % 4 == 0 and x % 8 != 0, lst))

print(lst)
```

## Ans of the Question  **2**


```python
def compute_ratio(x, y) :
  try :
    val = (x + y) / (x * y)
  except ZeroDivisionError :
    val = "Error: Division not possible"
  finally :
    return val

x,y = list(map(int, input().split()))

print(compute_ratio(x, y))
```

## Ans of the Question  **3**

```python
with open("./sample_data/data.txt") as file :
  lon = 0 ; lon_str = ""
  for x in file.read().split() :
    x = x.replace(',', '').replace('.', '').replace('?', '').replace('!', '').replace(';', '').replace("'", "")
    if(len(x) > lon) :
      lon = len(x)
      lon_str = x

  print(lon_str)
```

## Ans of the Question  **5**

```python
word = input().replace(" ", "").lower()

if word == word[::-1] :
  print("Palindrome")
else :
  print("Not Palindrome")
```
