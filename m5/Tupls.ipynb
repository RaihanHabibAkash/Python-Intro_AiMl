# Tuple

### Decleare

```python

tup = (10,20,30)
mix = (10, 10.28, "Akash", False)
print(tup, type(tup))
print(mix)
```

### Single val is not Tuple

```python
# Single Data is not tupe
tup = (10)
print(tup, type(tup))

# Converting into tupe
tup1 = tuple([10]) # list
tup2 = (10,)

print(tup1, type(tup1))
print(tup2, type(tup2))
```

### List into Tuple

```python
lst = [10,20,30]

tup1 = tuple(lst)
tup2 = tuple([10])

print(tup1, type(tup1))
print(tup2, type(tup2))
```

### Access & Slicing

```python
tup = (10,2.4, "Ash", False)

print(tup[0])
print(tup[3])


# Slice
new_tup = tup[0:3]
print(new_tup, type(new_tup))
```

### Tuple method

```python
tup = (20,10,30,10,40,10,10)

print(tup.count(10))
print(tup.index(30))
```
