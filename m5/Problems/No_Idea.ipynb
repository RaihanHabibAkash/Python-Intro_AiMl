### https://www.hackerrank.com/challenges/no-idea/problem

```python
inp = input().split()
n = int(inp[0]) ; m = int(inp[1])

arr = input().split()

a = set(input().split())
b = set(input().split())

sum = 0

for x in arr :
    if x in a :
        sum += 1
    if x in b :
        sum -= 1
print(sum)
```
