## Class & Object

```python
class Phone :
    category = "Elctric"
    def __init__(self, model, battery, camera, battery_per=100) :
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_per = battery_per

    def charge(self, hour) :
        print(f"Charge completed by {hour} {self.model}")
    def capture(self, photo) :
        if self.battery < photo :
            print("No Charge")
        else :
            self.battery -= photo
            print(f"Photo captured in {self.model}")

class Colling_Mechanism :
    def __init__(self, cooling_method) :
        self.cooling_method = cooling_method
    def cooling_on(self) :
        print(f"the system is being cool by {self.cooling_method}")
```

```python
apple = Phone("Iphone18", 2000,40)
samsung = Phone("s-23",4000,30)

apple.capture(1005)

print(apple.battery, samsung.battery)
```

## Single Inheritance

```python
class SmartPhone(Phone) :
    def __init__(self, model, battery, camera, processor) :
        super().__init__(model, battery, camera)
        self.processor = processor
    def charge(self, hour):
        print("Fast charging")
        super().charge(hour)
```

```python
pro = SmartPhone("X", 4000, 200, "Intel")
print(pro.model)
pro.charge(1)
```

## Multiple Inheritance

```python
class SmartPhn_Colling(SmartPhone, Colling_Mechanism) :
    def __init__(self,model, battery, camera,processor,cooling_method) :
        SmartPhone.__init__(self, model, battery, camera,processor)
        Colling_Mechanism.__init__(self, cooling_method)
```

```python
pro_col = SmartPhn_Colling("Y", 5000, 300, "SnapDragon", "Cool")

print(pro_col.processor)
print(pro_col.camera)
print(pro_col.cooling_method)
pro_col.cooling_on()
pro_col.charge(1)
```

## Poly morphism

```python
class Camera :
    def __init__(self, name) :
        self.name = name
    def capture(self) :
        print("A photo is captured")

class Smart_Phn(Camera) :
    def __init__(self, name, resulation) :
        super().__init__(name)
        self.resulation = resulation
    # Overwrite
    def capture(self) :
        print("Photo is clicked by PHN")

phn = Smart_Phn("Iphn", 30)
phn.capture()
```

## En Capsulation

```python
class Car :
    def __init__(self, name, col, id) :
        self.__name = name
        self.col = col
        self.__id = id
    def id_getter(self) :
        return self.__id
    def name_setter(self, name) :
        self.__name = name
    def name_getter(self) :
        return self.__name

lambo = Car("Lambo", "Red", 22)

print(lambo.col)
print(lambo.id_getter()) # Printing private id
lambo.name_setter("Far")
print(lambo.name_getter()) # Private id update
```
