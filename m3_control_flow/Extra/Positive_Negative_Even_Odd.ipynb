```python
n = int(input())
lst = list(map(int, input().split()))

pos = 0 ; neg = 0 ; evn = 0 ; odd = 0

for x in lst :
    if x > 0 :
        pos += 1
    elif x < 0 :
        neg += 1
    if x % 2 == 0 :
        evn += 1
    else :
        odd += 1
print(f"Positive: {pos}")
print(f"Negative: {pos}")
print(f"Even: {pos}")
print(f"Positive: {pos}")
```
