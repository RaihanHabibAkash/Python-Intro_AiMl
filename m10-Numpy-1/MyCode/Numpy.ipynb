```python
import numpy as np

# 1D Array
arr = np.array([1,2,3,4])
print(type(arr))

# 2D Array
arr2 = np.array([
    [1,2,3,4], # row1
    [4,5,6,7] # row2
])

# 3D Array
arr3 = np.array([
    # 1st floor
    [ [1,2,3,4],
        [4,5,6,7] ],
    # 2nd floor
    [ [1,2,3,4],
        [4,5,6,7] ],
    # 3rd floor
    [ [1,2,3,4],
        [4,5,6,7] ]
])

print(arr)
print(arr2)
print(arr3)
```

## Array Attributes

```python
# dimension
print(arr.ndim, arr2.ndim, arr3.ndim)

# shape
print(arr.shape, arr2.shape, arr3.shape)

# data type
print(arr.dtype, arr2.dtype, arr3.dtype)

# size
print(arr.size, arr2.size, arr3.size)
```

## Upcasting

```python
# Upcasting to float
arr = np.array([4,5,3.4])
print(arr.dtype)

# Upcasting to string
arr = np.array([3,4.4, "h"])
print(arr.dtype)
```

## Selecting a data type for an array

```python
arr = np.array([1,2,300], dtype=np.int16)
print(arr.dtype)

arr = arr.astype(np.int32)
print(arr.dtype)

# error
arr = np.array([1,2,3.4, "Hlw"], dtype = np.int32)
```

## Nd Array creation from existing data

```python
# List
lst = [10,20,30]
arr = np.array(lst, dtype = np.int32)
print(arr, type(arr), arr.dtype)


mixed_lst = [20, True, 4.5, "Hello"]
arr = np.array(mixed_lst)
print(arr, type(arr), arr.dtype)

matrix = [ [1,2,3], # row1
           [4,5,6] #row2
                ]
arr = np.array(matrix, dtype=np.int32)

print("----For last---")
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# Tuple

tup = (10,29,39)

arr = np.array(tup, dtype=np.int32)
print(arr.dtype)
arr = arr.astype(np.int16)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# Set

st = {1,3,4}

# Creating Array of sets
arr = np.array(st)
print(st)

arr = np.array(list(st), dtype=np.int32)
arr = arr.astype(np.int16)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
```

```python
# Dictionary

dc = {'a' : 10, 'b' : 20, 'c' : 30}

keys = dc.keys()
vals = dc.values()
items = dc.items()
```

```python
# keys to nd arrray
arr = np.array(keys)
print(arr.dtype)

arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#values to ndarray
arr = np.array(list(vals))
print(arr)
arr = arr.astype(np.int16)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

## array creation with random values

```python
# np.random.rand(shape)

arr = np.random.rand(3,4) # 0->1

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.randint(start,end, shape)
arr = np.random.randint(1,10, (2,4))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.random.uniform(start,end, shape)

arr = np.random.uniform(60, 100, (3,5))

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

## Array creation with range functions

```python
# np.arange(first , end , step ) -> similar to range function

arr = np.arange(1, 10, 1)
mat = arr.reshape(3,3)

print(arr)
print(mat)
```

```python
# np.linspace(start , end(included) , koita point)

arr = np.linspace(0, 4, 6)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start , end(included) , koita point)

arr = np.logspace(0, 4, 5) # base = 10(default)
print(arr)

arr = np.logspace(0,4, 6, base = 2) # 2^0, 2^0.8, 2^16, 2^2.4, 2^3.2, 2^4
print(arr)
```

## Indexing & Slicing

```python
# indexing
arr1 = np.array([10,20,39])
arr2 = np.array([
    [50,60,70,80],
    [100,200,300,400]
])

arr1[2] = 100
print(arr1)

arr2[0][2]= 100
print(arr2)
```

```python
#slicing
#1d = arr[start:end:step]

arr1_mod = arr1[0:2].copy()
print(arr1_mod)
```

```python
#2d slicing
# arr[ row_start : row_end : step , column_start : column_end : step ]

#getting a row

row_0 = arr2[0:1 ,]

row_1= arr2[1: , ]

# print(row_0)
# print(row_1)


#getting a column

col_0 = arr2[:: ,0:1]
col_1 = arr2[::,1:2]

print(col_0)
print(col_1)

#getting a portion


portion = arr2[:: , 0:2]

print(portion)
```

## Advanced Indexing
