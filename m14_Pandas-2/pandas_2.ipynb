# apply function on df

```python
# .apply() function is used to apply a function to every value in a Series or along rows/columns of a DataFrame
df
```

```python
#min max scaling. scaled values -> 0-1

mn = df['Total Marks'].min() 
mx = df['Total Marks'].max() 


df['Scaled Marks'] = df['Total Marks'].apply(lambda x : (x-mn)/(mx-mn))

df
```

```python
# custom built function 

def grading_system(marks):

    if marks >=260:
        return 'A+' 
    elif marks>=250:
        return 'A' 
    else:
        return 'A-'


df['Grade'] = df['Total Marks'].apply(grading_system) 

df
```

```python
def marking_system(df): 
    a = df['Data Structure Marks'] * 2 
    b = df['Python Marks'] * 3 
    c = df['Algorithm Marks'] * 4

    return a+b+c 


# Proti row ar jnn kaj ta hoise
df['Exceptional Marks'] = df.apply(marking_system,axis=1) 
```

# datetime and time delta 

```python
df = pd.read_csv('student_completed_data.csv')
df
```

```python
# dtype = obj
df['EnrollmentDate']
```

```python
# dtype = DateTime64 a convert hoise
df['EnrollmentDate'] = pd.to_datetime(df['EnrollmentDate']) 

df['EnrollmentDate']
```

```python
# year ta ber kore col create
df['Enrollment Year'] = df['EnrollmentDate'].dt.year

# day ta ber kore col create
df['Enrollment Date'] = df['EnrollmentDate'].dt.day

df
```

```python
# dateTime64 a conver
df['FinishedDate'] = pd.to_datetime(df['FinishedDate']) 

df['Total time taken to finish']= df['FinishedDate'] - df['EnrollmentDate']

df
```

# Groupby

```python
import numpy as np

# naull set kora hoise koekta value
df.loc[df['FullName'] == 'Afsana Mimi', 'Total Marks'] = np.nan
df.loc[df['FullName'] == 'Sadia Chowdhury', 'Total Marks'] = np.nan
df.loc[df['FullName'] == 'Alif Rahman', 'Total Marks'] = np.nan
```

```python
df
```

```python
# instructor col ar base kore grp create kora hoise
instructor_group = df.groupby('Instructor')

instructor_group
```

```python
# obj a (key-dataframe) ase
for instructor , df_ins in instructor_group: 
    print(instructor)
    display(df_ins) 
```

```python
# Total min
print(df.min())

 # proti grp ar min ditase
instructor_group.min()
```

```python
instructor_group.describe()
```

```python
# instructor grp theke mean values gulo rakhbo
mean_marks = instructor_group['Total Marks'].mean()

mean_marks
```

```python
# null value pele, oi row ar Instructor ar value boshabo map kore
df['Total Marks']= df['Total Marks'].fillna(df['Instructor'].map(mean_marks)) 

df
```
