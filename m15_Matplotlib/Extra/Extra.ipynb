# Extra Problems

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Problem 1:

```python
df = pd.read_csv("Practice.csv")

df["Discounted_Price"] = df["Price"] + (df["Price"] * 10 / 100)

df
```

## Problem 2:

```python
# Count duplicates
print(df["EmployeeID"].duplicated().sum())

df.drop_duplicates(subset=["EmployeeID"], inplace=True)

df
```

## Problem 3:

```python
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())

df.dropna(subset=["Salary"], inplace=True)

df
```

## Problem 4:

```python
print(df["Marks1"].mean(), df["Marks1"].max())
print(df["Marks2"].mean(), df["Marks2"].max())
print(df["Marks3"].mean(), df["Marks3"].max())

Total_Marks = df["Marks1"] + df["Marks2"] + df["Marks3"]
print(Total_Marks)

Total_Marks.std()
```

## Problem 5:

```python
plt.xlabel('Age') 
plt.ylabel('Frequency of Students') 
plt.hist(df["Age"], color="Red", edgecolor='black')
```

```python
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("People of Age vs Salary")

plt.scatter(df["Age"], df["Salary"])
```

```python
plt.xlabel("Price")
plt.ylabel("Discounted_Price")
plt.title("Price vs Discounted_Price")

plt.scatter(df["Price"], df["Discounted_Price"])
```
