### 1 . Create 
a 1D, 2D, and 3D NumPy array and print their dimensions and shapes.

```python
arr1 = np.random.randint(1,10, (7,))
arr2 = np.random.randint(1, 20, (3, 5))
arr3 = np.random.randint(1, 100, (3,4,4))

print(arr1.shape, arr1.ndim)
print(arr2.shape, arr2.ndim)
print(arr3.shape, arr3.ndim)
```

### 2 . Create:

```python
arr1 = np.zeros((3,3))
arr2 = np.ones((2, 5))
arr3 = np.full((4,), 7)

print(arr1)
print(arr2)
print(arr3)
```

## 4. Create a 4×4 matrix from 0–15. Then , extract:
the first row


the last column


a 2×2 subarray from the center


```python
mat = np.arange(0, 16).reshape(4,4)

print(mat) 
print()

print(mat[:1])
print()

print(mat[:, 3:])
print()

print(mat[1:3, 1:3])
```
