# Dictionary

### Decleare

```python
dic = {}
print(type(dic))

dic = { "name" : "Akash", 1 : "roll", "numbers" : [10,20,30] }
print(dic, type(dic))
```

### Access

```python
dic = { "name" : "Akash", 1 : "roll", "numbers" : [10,20,30] }
print(dic["numbers"])
# print(dic.get("name"))
```

### Duplicates key

```python
dic = {"name" : "Akash", "roll" : 2, "height" : 5.7, "roll" : 20}
print(dic)
```

### Key ta na thakle

```python
dic = {"name" : "Akash", "roll" : 2, "height" : 5.7, "roll" : 20}
print(dic.get("lol"))
print(dic.get("lol", "Nai ga"))
```

### Update & Delete

```python
dic = {"name" : "Akash", "roll" : 2, "height" : 5.7, "roll" : 20}

dic["name"] = "Ai/Ml"
# dic.update({"name" : "Ai/Ml"})
print(dic)

del dic["name"]

print(dic)
```

### Imuteble jinish e rakha jabe key hisebe

```python
# dic = {[10,29] : "Akash"} # rakha jabe na list

dic = { (10,20,30) : "akash" }
print(dic)

dic["dic"] = {"roll" : 3, "height" : 3.44}
print(dic)
```

## Dictionary view methods

```python
dic = { "name" : "Akash", 1 : "roll", "numbers" : [10,20,30] }

keys = dic.keys() # Auto updates
print(keys)
dic["lol"] = "Bro"
print(keys)

values = dic.values() # Auto updates
print(values)
dic["lol"] = "ASH"
print(values)

items = dic.items()
print(items)

# Iteration
for key,value in dic.items() :
    print(key, value)
```

## Dictionary Comprehension

```python
square = { x:x**2 for x in range(1,11) if x%2==0 }
print(square)

co_ordinates = [(20,10), (11, 2.5), (12,2,4)]
loc = ["dhaka", "Sylhet", "Barishal"]

exect_loc = { lo : co_or for lo,co_or in zip(loc, co_ordinates) }
print(exect_loc)
```
