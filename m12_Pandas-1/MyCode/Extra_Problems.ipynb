## Problem-5:

## Problem-6:
Create a DataFrame of employees with columns "Name", "Department", and 
"Salary". Sort it by "Salary" in ascending order and by "Name" in descending 
order

```python
dirc = {
    "Name": ["Raihan", "Zap", "Habib", "Akash", "Bro"],
    "Department": ["Cse", "EEE", "Math", "English", "Software"],
    "Salary": [1e5, 1e2, 1e4, 1e5, 1e5]
}

df = pd.DataFrame(dirc)

df
```

```python
df.sort_values(["Salary", "Name"], ascending=[1, 0], inplace=True)

df
```

## Problem-7:
Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print 
students with Score > 60 and those with Score < 40

```python
tup = (
    ("Raihan", 89),
    ("Habib", 90),
    ("Bro", 33),
    ("Akash", 93),
    ("Ash", 23),
    ("Niloy", 93),
    ("Sadman", 37),
    ("Tony", 23)
)

df = pd.DataFrame(tup, columns=["Name", "Score"])

df
```

```python
score_above_60 = df.loc[df["Score"] > 60]

score_above_60
```

```python
score_down_40 = df.loc[df["Score"] < 40]

score_down_40
```
