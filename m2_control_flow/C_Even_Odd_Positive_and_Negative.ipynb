```python
cs = int(input())
inp = input()
inp = inp.split()

evn = 0 ; odd = 0 ; neg = 0 ; pos = 0

for i in range(cs) :
    x = int(inp[i])

    if(x < 0) :
        neg += 1
    elif(x > 0) :
        pos += 1

    if(x % 2 == 0) :
        evn += 1
    else :
        odd += 1

print("Even:", evn)
print("Odd:", odd)
print("Positive:", pos)
print("Negative:", neg)
```
