```python
# You must run this code and use this dataset for all questions

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

## Problem-5:
Using NumPy:

Assume all salaries of Tech roles increase by 8%

Calculate the new total salary expense

Find the overall increase in expense after the raise

Print both old and new totals along with the difference.

```python
new_arr = np.zeros_like(emp_roles, dtype=np.float64)

for i in range(0, emp_income.size):
  if((emp_roles[i] == "Software Developer") or (emp_roles[i] == "Backend Developer") or (emp_roles[i] == "Digital Marketer")) :
    new_arr[i] = emp_income[i] + (emp_income[i] * 8 / 100)
  else :
    new_arr[i] = emp_income[i]

old_totals = np.sum(emp_income)
new_totals = np.sum(new_arr)

print(old_totals)
print(new_totals)
print(new_totals - old_totals)
```

## Problem-10:
Using .loc:

Increase salary by 12% for "Software Developer" and "Backend Developer"

Increase salary by 5% for all other roles

Then:

Print the dataframe that includes role , previous_salary , new_salary , increase_of_salary

```python
dev_data = df.loc[(df["role"] == "Software Developer") | (df["role"] == "Backend Developer")]
dev_old_salary = dev_data["emp_salary"]
dev_new_salary = dev_old_salary + (dev_old_salary * 12 / 100)
dev_increased_salary = dev_new_salary - dev_old_salary

others_data = df.loc[(df["role"] != "Software Developer") & (df["role"] != "Backend Developer")]
others_old_salary = others_data["emp_salary"]
others_new_salary = others_old_salary + (others_old_salary * 5 / 100)
others_increased_salary = others_new_salary - others_old_salary

total_new_salary = pd.concat([dev_new_salary, others_new_salary]).sort_index()
total_increased_salary = pd.concat([dev_increased_salary, others_increased_salary]).sort_index()


df["old_salary"] = df["emp_salary"]
df["new_salary"] = total_new_salary
df["increase_of_salary"] = total_increased_salary
df.drop(columns=["emp_id", "emp_age", "emp_salary"], inplace=True)

df
```
